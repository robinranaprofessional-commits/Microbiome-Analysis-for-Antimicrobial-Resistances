After analyzing the dataset, we noticed that the ID's were labelled as "IA" and "IE" based on the time period they were taken removing the IA and IE using str[2:] allowed us to determine which entries were analyzed twice, this can be helpful in determining which 

```python
first_ids = set(individual_samples[individual_samples["Time"] == "1st time point"]["ID"].str[2:])

second_ids = set(individual_samples[individual_samples["Time"] == "2nd time point"]["ID"].str[2:])

print("Matching numbers:", len(first_ids & second_ids))
print("Only in first:", first_ids - second_ids)
print("Only in second:", second_ids - first_ids)
```

Above, we have listed all individual samples as well as organized them based on when they were tested. 
32 of the individual samples were tested twice, once in each time period, we also have 15 samples in both the 1st and 2nd time point that were only tested once. 


```python
individual_samples["Animal_ID"] = individual_samples["ID"].str[2:]
print(individual_samples["Animal_ID"].nunique())
```
